```python
from scipy.stats import ttest_ind

exam1 = [78, 85, 90, 88, 75, 95, 89]
exam2 = [70, 80, 85, 79, 72, 82, 81]

t_stat, p_value = ttest_ind(exam1, exam2)

print("T-statistic =", t_stat)
print("P-value =", p_value)

if p_value < 0.05:
    print("Significant difference")
else:
    print("No significant difference")
```

```python
import numpy as np
from scipy import signal
import matplotlib.pyplot as plt

t = np.linspace(0,1,500)

clean = np.sin(2*np.pi*5*t)
noise = np.random.normal(0,0.5,500)

noisy = clean + noise

b, a = signal.butter(4, 0.1)

filtered = signal.filtfilt(b,a,noisy)

plt.plot(t,noisy,label="Noisy")
plt.plot(t,filtered,label="Filtered")
plt.legend()
plt.show()
```

```python
# MINI PROJECT: Employee Data Analysis Using SciPy

import pandas as pd
import numpy as np
from scipy.stats import pearsonr, linregress, ttest_ind

# Sample Employee Dataset
df = pd.DataFrame({
    "Age": [22, 25, 28, 30, 35, 38, 40, 45, 48, 50],
    "Salary": [25000, 30000, 35000, 40000, 50000, 55000, 60000, 70000, 75000, 80000],
    "Experience": [1, 2, 3, 5, 7, 8, 10, 12, 14, 15],
    "Performance_Score": [60, 65, 68, 72, 78, 80, 85, 88, 90, 95]
})

print("EMPLOYEE DATASET")
print(df)

# --------------------------------------------------
# 1. Correlation Analysis
# --------------------------------------------------

print("\n----- CORRELATION ANALYSIS -----")

corr_sal_exp, p_sal_exp = pearsonr(df["Salary"], df["Experience"])
corr_perf_exp, p_perf_exp = pearsonr(df["Performance_Score"], df["Experience"])
corr_sal_perf, p_sal_perf = pearsonr(df["Salary"], df["Performance_Score"])

print(f"Salary vs Experience Correlation = {corr_sal_exp:.4f}")
print(f"Performance vs Experience Correlation = {corr_perf_exp:.4f}")
print(f"Salary vs Performance Correlation = {corr_sal_perf:.4f}")

# --------------------------------------------------
# 2. Linear Regression
# --------------------------------------------------

print("\n----- LINEAR REGRESSION -----")

reg = linregress(df["Experience"], df["Salary"])

print(f"Slope = {reg.slope:.2f}")
print(f"Intercept = {reg.intercept:.2f}")
print(f"R-squared = {reg.rvalue**2:.4f}")

print(f"\nRegression Equation:")
print(f"Salary = {reg.slope:.2f} * Experience + {reg.intercept:.2f}")

# --------------------------------------------------
# 3. Statistical Test (T-Test)
# --------------------------------------------------

print("\n----- T-TEST -----")

young_group = df[df["Age"] < 35]["Performance_Score"]
senior_group = df[df["Age"] >= 35]["Performance_Score"]

t_stat, p_value = ttest_ind(young_group, senior_group)

print(f"T-statistic = {t_stat:.4f}")
print(f"P-value = {p_value:.4f}")

if p_value < 0.05:
    print("Result: Significant difference between groups")
else:
    print("Result: No significant difference between groups")

# --------------------------------------------------
# 4. Significant Relationships
# --------------------------------------------------

print("\n----- SIGNIFICANCE TESTS -----")

relationships = {
    "Salary vs Experience": p_sal_exp,
    "Performance vs Experience": p_perf_exp,
    "Salary vs Performance": p_sal_perf
}

for relation, p in relationships.items():
    if p < 0.05:
        print(f"{relation} --> Statistically Significant (p={p:.5f})")
    else:
        print(f"{relation} --> Not Significant (p={p:.5f})")

# --------------------------------------------------
# 5. Conclusions
# --------------------------------------------------

print("\n----- CONCLUSIONS -----")

if corr_sal_exp > 0:
    print("✓ Salary increases as Experience increases.")

if corr_perf_exp > 0:
    print("✓ Performance improves with Experience.")

if corr_sal_perf > 0:
    print("✓ Higher-performing employees tend to earn higher salaries.")

if reg.rvalue**2 > 0.8:
    print("✓ Experience is a strong predictor of Salary.")

print("✓ Employee experience has a major influence on salary and performance.")
```
